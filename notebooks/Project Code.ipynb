# Module Code to determine if the person has diabetes or not / the should take the mean value throuogh months record

```python
def diabetes(FBS, HbA1C):
    if FBS < 70:
        return "Low blood sugar, Consult your doctor"
    elif 70 <= FBS <= 99:
        if HbA1C < 5.7:
            return "Normal blood sugar and Normal HbA1C"
        else:
            return "Normal FBS, HbA1C indicates a risk of diabetes."
    elif 100 <= FBS <= 125:
        if 5.7 <= HbA1C <= 6.4:
            return "Prediabetes. Lifestyle need to changed as recommended."
        else:
            return "Prediabetes based on FBS."
    elif FBS >= 126 or HbA1C >= 6.5:
        return "Diabetes. Consult Doctor."
    else:
        return "Invalid input or unclassified result."

# Please enter the values in diabetes(FBS, HbA1C)
result = diabetes(110, 5.9)
print(result)
```

# Pie Plot / Patient Sex

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
vdf= pd.read_csv("NNRaw_Data.csv")


patient_counts = vdf['PatientSex'].value_counts()

# Seaborn color palette
colors = sns.color_palette('Set2', n_colors=len(patient_counts))

plt.figure(figsize=(8, 8), dpi=120)

plt.pie(
    patient_counts,
    labels=patient_counts.index,
    autopct='%1.1f%%',  
    startangle=140,  # Rotate the angle 
    colors=colors,
    wedgeprops={'edgecolor': 'white', 'linewidth': 1.5, 'linestyle': 'solid'}, 
    pctdistance=0.85)

# Add a title
plt.title('Dataset Distribution of Patient Sex', fontsize=16, weight='bold')

# Add donut chart 
centre_circle = plt.Circle((0, 0), 0.70, color='white', fc='white', linewidth=1)
plt.gca().add_artist(centre_circle)

# Add legend 
plt.legend(
    patient_counts.index,
    title="Patient Sex Percentage ",
    loc="upper left",
    fontsize=12,
    title_fontsize=14,
    labels=[f'{label}: {count}' for label, count in zip(patient_counts.index, patient_counts)]
)

plt.tight_layout()
plt.savefig('Patient Sex Percentage.jpg', dpi=300, bbox_inches='tight')

plt.show()
```

# Regression Line Plot / HBA1C vs Creatinine

```python

import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
vdf= pd.read_csv("NNRaw_Data.csv")


sns.set(style="whitegrid")

# Regression plot
plt.figure(figsize=(10, 6))
sns.regplot(
    data=vdf,
    x='HBA1CResult',
    y='CreatinineResult',
    scatter_kws={'color': 'cyan', 's': 50, 'edgecolor': 'black'}, 
    line_kws={'color': 'red', 'linewidth': 1},  
    ci=95  # Confidence interval
)
plt.title('Regression Line Plot: HBA1C vs Creatinine', fontsize=14)
plt.xlabel('HBA1C', fontsize=12)
plt.ylabel('Creatinine', fontsize=12)

plt.tight_layout()
plt.savefig('HBA1C vs Creatinine.jpg', dpi=300, bbox_inches='tight')

plt.show()
```
